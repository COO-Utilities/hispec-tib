# Example OTA


Read https://coo-utilities.github.io/hispec-tib/ota.html

```python
from pathlib import Path
import sys

TOOLS = next((p.resolve()
              for root in (Path.cwd(), *Path.cwd().parents)
              for p in (root, root / 'tools', root / 'hispec-tib/tools')
              if (p / 'hispec_fibpcb.py').is_file()), None)
if TOOLS is None:
    raise RuntimeError('Open this notebook from the repository or its tools directory.')
if str(TOOLS) not in sys.path:
    sys.path.insert(0, str(TOOLS))
import hispec_fibpcb as hspcb
```

# Connect

```python
pcb = hspcb.HispecFibPcb('hispec.caltech.edu', device='hsfib-tib')
```

## Force Lasers Off

```python
pcb.laser_bankpower('override_off')
```

## Go use the helper to OTA it

```python
pcb.ota()
```

## Give the all clear

```python
candidate = pcb.ota()
image_hash = candidate.image_hash
print(f"Image {image_hash}: confirmed={candidate.confirmed}, remaining={candidate.trial_remaining_s}s")
```

```python
# Do any check you want here. Board is alive and talking at least!

```

```python
# Run only after checking the candidate; this accepts the flashed image
pcb.confirm_image(image_hash)
```

## Back to normal ops

```python
pcb.laser_bankpower('auto')
```
